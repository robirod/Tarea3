**AquaLimpia S. A. Analisis desempeño plantas de tratamiento **

## Objetivos del Proyecto

### General
- Desarrollo colaborativo en base a AquaLimpia, del cual desarrollar dashboard exploratorios por medio de Python e identificar comportamiento del tratamiento operacional y gestión ambiental

### Específicos
- Identificar patrones en los niveles DBO y eficiencia de tratamiento.

- Verificar cumplimiento de normas por plantas

- Generar archivos de salida específicos para las áreas para area de operaciones y gestión ambiental

## Procesos

1. **Carga y Limpieza de Datos:** Importación del dataset oficial y tratamiento de valores nulos o inconsistencias.
2. **Análisis Exploratorio:** Visualización de tendencias de caudal, carga contaminante.
3. **Segmentación por Áreas:**
    - **Operaciones:** Análisis de procesamiento de entradas y salidas de DBO y comportamiento de caudal de entrada y lodos generados.
    - **Gestión Ambiental:** Monitoreo de cumplimiento normativo y tendencias de calidad.
4. **Generación de Entregables:** Exportación de archivos CSV/Excel personalizados para cada departamento.

## Resultados

La visualización de graficos en dashboard exploratorio presenta una relación en el area operacional del cual a mayor DBO de entrada mayor sera DBO de salidas, al igual que mayor caudal de entrada presenta mayor caudal de salida. Junto con esto, la grafica de gestión ambiental indica que a mayor DBO de salida, son menores los casos de cumplimiento de normas, por lo que es importante detectar de manera temprana la carga contaminante de entrada para poder gestionar estos en relación al cumplimiento de normas ambientales.
